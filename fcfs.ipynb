```python

processes = [
    ["P1", 0, 2],
    ["P2", 2, 3],
    ["P3", 3, 3],
    ["P4", 5, 2]
]

processes.sort(key=lambda x: x[1])

n = len(processes)
completion_time = [0] * n
turnaround_time = [0] * n
waiting_time = [0] * n

current_time = 0

for i in range(n):
    pid = processes[i][0]
    arrival = processes[i][1]
    burst = processes[i][2]

    if current_time < arrival:
        current_time = arrival

    current_time += burst
    completion_time[i] = current_time

    turnaround_time[i] = completion_time[i] - arrival
    waiting_time[i] = turnaround_time[i] - burst

print(f"{'PID':<5}{'AT':<5}{'BT':<5}{'CT':<5}{'TAT':<5}{'WT':<5}")

total_wt = 0
total_tat = 0

for i in range(n):
    pid = processes[i][0]
    arrival = processes[i][1]
    burst = processes[i][2]

    print(f"{pid:<5}{arrival:<5}{burst:<5}{completion_time[i]:<5}{turnaround_time[i]:<5}{waiting_time[i]:<5}")

    total_wt += waiting_time[i]
    total_tat += turnaround_time[i]

print(f"\nAverage Waiting Time: {total_wt/n:.2f}")
print(f"Average Turnaround Time: {total_tat/n:.2f}")
```
