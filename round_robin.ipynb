```python
n= int(input("enter number of process"))
brust= []
arrive = []
for i in range(n):
  bt = int(input(f"enter brust of {i+1}:"))
  brust.append(bt)

  arrive.append(0)
print(brust)
print(arrive)
quantum = int(input("enter quantum"))
remaining = brust.copy()
completion = [0]*n
turnaround = [0]*n
waiting = [0]*n
time = 0
queue =list(range(n))
while queue:
  i = queue.pop(0)
  if remaining[i] > quantum:
    time += quantum
    remaining[i] -= quantum
    queue.append(i)
  else:
    time += remaining[i]
    remaining[i] = 0
    completion[i] = time
    turnaround[i] = completion[i] - arrive[i]
    waiting[i] = turnaround[i] - brust[i]
print("\nP\tBT\tCT\tTAT\tWT")
total_wt = 0
total_tat = 0

for i in range(n):
    print(f"P{i+1}\t{brust[i]}\t{completion[i]}\t{turnaround[i]}\t{waiting[i]}")
    total_wt += waiting[i]
    total_tat += turnaround[i]

print(f"\nAverage Waiting Time = {total_wt/n:.2f}")
print(f"\nAverage Turnaround Time = {total_tat/n:.2f}")




```
