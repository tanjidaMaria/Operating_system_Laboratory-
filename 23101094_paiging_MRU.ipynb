```python
referrence = [7, 0, 1, 2, 0, 3, 0, 4, 2, 3]
n = 4
mem = []
faults = 0
for p in referrence:
  if p in mem:
    mem.remove(p)


  else:
    faults += 1
    if len(mem)==n:
      mem.pop()
  mem.append(p)
print(" MRU fault:", faults)
```
