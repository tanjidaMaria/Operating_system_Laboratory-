```python
referrence = [1, 2, 3,4,1,2,3,1]
n = 3
mem = []
faults = 0
for p in referrence:
  if p not in mem:
    faults += 1
    if len(mem)==n:
      mem.pop(0)
    mem.append(p)
print("Fifo referrence fault:", faults)
```
