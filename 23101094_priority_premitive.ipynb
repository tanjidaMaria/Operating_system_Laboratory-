```python
processes = [
    ["P1", 3, 4, 2],
    ["P2", 4, 2, 0],
    ["P3", 2, 6, 3],
    ["P4", 0, 3, 6],
    ["P5", 1, 2, 1],

]

n = len(processes)
remaining_time = [processes[i][2] for i in range(n)]
completion_time = [0] * n
turnaround_time = [0] * n
waiting_time = [0] * n
completed = [False] * n

current_time = 0
done = 0

while done < n:
    idx = -1
    priority = 9999

    for i in range(n):
        if processes[i][1] <= current_time and completed[i] == False and remaining_time[i] > 0:
            if processes[i][3] < priority:
                priority = processes[i][3]
                idx = i

    if idx == -1:
        current_time += 1
    else:
        remaining_time[idx] -= 1
        current_time += 1

        if remaining_time[idx] == 0:
            completed[idx] = True
            done += 1

            arrival = processes[idx][1]
            burst = processes[idx][2]

            completion_time[idx] = current_time
            turnaround_time[idx] = completion_time[idx] - arrival
            waiting_time[idx] = turnaround_time[idx] - burst

print(f"{'PID':<5}{'AT':<5}{'BT':<5}{'CT':<5}{'TAT':<5}{'WT':<5}")

total_wt = 0
total_tat = 0

for i in range(n):
    print(f"{processes[i][0]:<5}{processes[i][1]:<5}{processes[i][2]:<5}{completion_time[i]:<5}{turnaround_time[i]:<5}{waiting_time[i]:<5}")
    total_wt += waiting_time[i]
    total_tat += turnaround_time[i]

print(f"\nAverage Waiting Time: {total_wt/n:.2f}")
print(f"Average Turnaround Time: {total_tat/n:.2f}")
```
