Non premative

```python
# Non-Preemptive SJF Scheduling Algorithm

processes = [
    ["P1", 3, 2],
    ["P2", 2, 5],
    ["P3", 3, 3],
    ["P4", 6, 4],
    ["P5", 2, 9],
    ["P6", 1, 5]
]

n = len(processes)
completion_time = [0] * n
turnaround_time = [0] * n
waiting_time = [0] * n
completed = [False] * n

current_time = 0
done = 0

while done < n:
    idx = -1
    min_burst = 9999

    for i in range(n):
        if processes[i][1] <= current_time and completed[i] == False:
            if processes[i][2] < min_burst:
                min_burst = processes[i][2]
                idx = i

    if idx == -1:
        current_time += 1
    else:
        arrival = processes[idx][1]
        burst = processes[idx][2]

        current_time += burst
        completion_time[idx] = current_time
        turnaround_time[idx] = completion_time[idx] - arrival
        waiting_time[idx] = turnaround_time[idx] - burst

        completed[idx] = True
        done += 1

print(f"{'PID':<5}{'AT':<5}{'BT':<5}{'CT':<5}{'TAT':<5}{'WT':<5}")

total_wt = 0
total_tat = 0

for i in range(n):
    print(f"{processes[i][0]:<5}{processes[i][1]:<5}{processes[i][2]:<5}{completion_time[i]:<5}{turnaround_time[i]:<5}{waiting_time[i]:<5}")
    total_wt += waiting_time[i]
    total_tat += turnaround_time[i]

print(f"\nAverage Waiting Time: {total_wt/n:.2f}")
print(f"Average Turnaround Time: {total_tat/n:.2f}")
```

Premative

```python
# Preemptive SJF (SRTF) Scheduling Algorithm

processes = [
    ["P1", 3, 2],
    ["P2", 2, 5],
    ["P3", 3, 3],
    ["P4", 6, 4],
    ["P5", 2, 9],
    ["P6", 1, 5]
]

n = len(processes)
remaining_time = [processes[i][2] for i in range(n)]
completion_time = [0] * n
turnaround_time = [0] * n
waiting_time = [0] * n
completed = [False] * n

current_time = 0
done = 0

while done < n:
    idx = -1
    min_remaining = 9999

    for i in range(n):
        if processes[i][1] <= current_time and completed[i] == False and remaining_time[i] > 0:
            if remaining_time[i] < min_remaining:
                min_remaining = remaining_time[i]
                idx = i

    if idx == -1:
        current_time += 1
    else:
        remaining_time[idx] -= 1
        current_time += 1

        if remaining_time[idx] == 0:
            completed[idx] = True
            done += 1

            arrival = processes[idx][1]
            burst = processes[idx][2]

            completion_time[idx] = current_time
            turnaround_time[idx] = completion_time[idx] - arrival
            waiting_time[idx] = turnaround_time[idx] - burst

print(f"{'PID':<5}{'AT':<5}{'BT':<5}{'CT':<5}{'TAT':<5}{'WT':<5}")

total_wt = 0
total_tat = 0

for i in range(n):
    print(f"{processes[i][0]:<5}{processes[i][1]:<5}{processes[i][2]:<5}{completion_time[i]:<5}{turnaround_time[i]:<5}{waiting_time[i]:<5}")
    total_wt += waiting_time[i]
    total_tat += turnaround_time[i]

print(f"\nAverage Waiting Time: {total_wt/n:.2f}")
print(f"Average Turnaround Time: {total_tat/n:.2f}")
```
