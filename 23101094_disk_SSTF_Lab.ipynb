```python
req_seq = [0, 13,41,50,63,89,100]
head =46
pending = req_seq.copy()
totalseek = 0
while pending:
  closest = min(pending, key=lambda i: abs(i-head))
  totalseek += abs(closest-head)
  head = closest
  pending.remove(closest)
print("total seek", totalseek)
```
